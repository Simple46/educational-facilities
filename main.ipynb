```python
df.drop(columns=["Form Hub Photo ID", "GPS", "Survey ID", "Facility ID" ], inplace=True)
df.head()
```

```python
# changing to date data type
df["Date Of Survey"] = pd.to_datetime(df["Date Of Survey"], errors="coerce")
df.dtypes
```

```python
# drop the facilities that is NaN
df.dropna(subset=["Facility Name"], inplace=True)
df["Facility Name"].head(30)
df["Facility Name"].isna().sum()

```

```python
df.dropna(subset=["Facility Type Display", "Improved Water Supply", "Improved Sanitation", "Chalkboard in Each Classes (T/F)"], inplace=True)
df.isna().sum()
```

```python
#replace num of teacher with nce and num of full time teacher to 0

df[["Num of Teachers With NCE", "Num Of Fulltime Teachers"]] = df[["Num of Teachers With NCE", "Num Of Fulltime Teachers"]].fillna(0)
df.isna().sum()
```

```python
# get the total number of student by adding the male with female student
df["Total Number Of Students"] = df["Number Of Male Students"] + df["Number Of Female Students"]
df.isna().sum()
```

```python
# now drop remaining nan number of stdents
df.dropna(subset=["Total Number Of Students", "Total Number Of Classrooms", "Total Number Of Toilets"], inplace=True)
df.isna().sum()
```

```python
# drop the remaining nan rows 
df = df.dropna()
```

```python
df
```

```python
df =  df[df["Management"] != "none"]
df["Management"].value_counts()
# df["Management"].isna().sum()



```

```python
df.isna().sum()
```

```python
df = df.dropna()
df.isna().sum()
```

```python
df.columns.tolist()
```

```python
#create column for total teachers
df["Total Number Of Teachers"] = (
    df["Number Of Male Teachers"] + df["Number Of Female Teachers"]
)


#create column for student to teacher ratio
df["Student Teacher Ratio"] = np.where(
    df['Total Number Of Teachers'] > 0,
    df["Total Number Of Students"] / df["Total Number Of Teachers"],
    np.nan
)
```

```python
df[["Total Number Of Students", "Total Number Of Teachers", "Student Teacher Ratio"]].head()
```

```python
#Creating the facility Density column
df["Facility Density"] = df.groupby("Community")["Facility Name"].transform("count")
```

```python
df.head()
```

```python
# creating a seperate column for states and LGA 
df[["State", "Lga"]] = df["Unique LGA"].str.split("_", n=1, expand=True)

df.head()
```

```python
# save the clean dataset
df.to_csv("data/Cleaned_dataset.csv", index=False)
```

```python
public_private_students = (
       df[df["Management"].isin(["public", "private"])].groupby("Management")["Total Number Of Students"]
    .mean()
    .reset_index()
)

public_private_students
```

```python
public_private_teachers = (
    df[df["Management"].isin(["public", "private"])]
    .groupby("Management")["Total Number Of Teachers"]
    .mean()
    .reset_index()
)

public_private_teachers
```

```python
fig4 = px.bar(
    public_private_teachers,
    x="Management",
    y="Total Number Of Teachers",
    title="Average Teacher Count: Public vs Private Schools",
    labels={
        "Management": "Management Type",
        "Total Number Of Teachers": "Average Teachers"
    },
    text="Total Number Of Teachers"
)

fig4.update_traces(
    texttemplate="%{text:.0f}",
    textposition="outside"
)

fig4.update_layout(
    height=500,
    title_x=0.5
)

fig4.show()
```

```python
# Infrastructure Comparison
infrastructure = (
    df[df["Management"].isin(["public", "private"])]
    .groupby("Management")[
        [
            "Improved Water Supply",
            "Improved Sanitation",
            "PHCN Electricity"
        ]
    ]
    .mean()
    .mul(100)
    .reset_index()
)

infrastructure
```

```python
infrastructure_plot = infrastructure.reset_index()

infrastructure_plot = infrastructure_plot.melt(
    id_vars="Management",
    var_name="infrastructure",
    value_name="percentage"
)
```

```python
fig5 = px.bar(
    infrastructure_plot,
    x="Management",
    y="percentage",
    color="infrastructure",
    barmode="group",
    title="Infrastructure Availability: Public vs Private",
    labels={
        "Management": "Management Type",
        "percentage": "Percentage of Schools",
        "infrastructure": "Infrastructure"
    }
)

fig5.show()
```

```python
# Top 10 State with Lowest Electricity

electricity_by_state = (
    df.groupby("State")["PHCN Electricity"]
    .mean()
    .mul(100)
    .reset_index()
)

electricity_by_state.columns = [
    "State",
    "electricity_percentage"
]

electricity_by_state.sort_values(
    "electricity_percentage"
).head(10)
```

```python
fig7 = px.bar(
    electricity_by_state.sort_values(
        "electricity_percentage"
    ).head(10),
    x="electricity_percentage",
    y="State",
    orientation="h",
    title="10 States with Lowest Electricity Access",
    labels={
        "electricity_percentage": "Schools with Electricity (%)"
    }
)

fig7.show()
```

```python
# States with the Lowest Sanitation Access
sanitation_by_state = (
    df.groupby("State")["Improved Sanitation"]
    .mean()
    .mul(100)
    .reset_index()
)

sanitation_by_state.columns = [
    "State",
    "sanitation_percentage"
]

sanitation_by_state.sort_values(
    "sanitation_percentage"
).head(10)
```

```python
fig8 = px.bar(
    sanitation_by_state.sort_values(
        "sanitation_percentage"
    ).head(10),
    x="sanitation_percentage",
    y="State",
    orientation="h",
    title="10 States with Lowest Improved Sanitation Access",
    labels={
        "sanitation_percentage": "Schools with Improved Sanitation (%)"
    }
)

fig8.show()
```

```python
# Create a reusable function for creating the pue chart
def create_pie_chart(column, title):

    counts = df[column].value_counts(dropna=False)

    fig = px.pie(
        values=counts.values,
        names=counts.index,
        title=title
    )

    fig.show()
```

```python
# improved water Supply
create_pie_chart(
    "Improved Water Supply",
    "Improved Water Supply"
)
```

```python
# Improved Sanitation
create_pie_chart(
    "Improved Sanitation",
    "Improved Sanitation"
)
```

```python
# Electricity Availability
create_pie_chart(
    "PHCN Electricity",
    "Electricity Availability"
)
```
