```python
# Convert the 3 columns to boolean

boolean_columns = [
    "improved_water_supply",
    "improved_sanitation",
    "phcn_electricity",
    "chalkboard_each_classroom_yn"
]

for column in boolean_columns:
    df[column] = df[column].astype("boolean")
```

```python
df.dtypes
```

```python
# Deal with missing values

missing = pd.DataFrame({
    "Missing": df.isna().sum(),
    "Percentage": (df.isna().sum()/ len(df)) * 100
})

missing = missing[missing["Missing"] > 0].sort_values(
    "Percentage",
    ascending=False
)

missing
```
